```python
#  ****CORRECT RECORD RETRIEVAL CODE - all spider-man movies****

import requests
import pandas as pd

url = "https://files.tmdb.org/p/exports/movie_ids_10_20_2026.json.gz"

df = pd.read_json(url, lines=True)

response = requests.get(
    url,
    headers={"User-Agent": "WebDataScience/1.0 (assh8422@colorado.edu)"},
)

print("HTTP Status Code:", response.status_code)
response.raise_for_status()


spiderman = df[df["original_title"].str.contains("Spider-Man:")]

print(spiderman[["id", "original_title", "popularity"]])
```
